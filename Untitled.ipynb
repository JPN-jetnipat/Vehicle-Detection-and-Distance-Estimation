```python
import os
import sys
import platform
import subprocess
import multiprocessing

print("=" * 80)
print("SYSTEM INFORMATION")
print("=" * 80)

print(f"Hostname           : {platform.node()}")
print(f"OS                 : {platform.platform()}")
print(f"Python             : {platform.python_version()}")
print(f"Python Executable  : {sys.executable}")
print(f"Working Directory  : {os.getcwd()}")
print(f"CPU Cores          : {multiprocessing.cpu_count()}")

print("\n" + "=" * 80)
print("MEMORY")
print("=" * 80)
os.system("free -h")

print("\n" + "=" * 80)
print("DISK")
print("=" * 80)
os.system("df -h")

print("\n" + "=" * 80)
print("NVIDIA GPU (PCI)")
print("=" * 80)
os.system("lspci | grep -i nvidia")

print("\n" + "=" * 80)
print("NVIDIA DRIVER")
print("=" * 80)
os.system("cat /proc/driver/nvidia/version")

print("\n" + "=" * 80)
print("NVIDIA DEVICE FILES")
print("=" * 80)
os.system("ls -l /dev/nvidia*")

print("\n" + "=" * 80)
print("NVIDIA-SMI")
print("=" * 80)
os.system("nvidia-smi")

print("\n" + "=" * 80)
print("CUDA TOOLKIT")
print("=" * 80)
os.system("nvcc --version")

print("\n" + "=" * 80)
print("PYTORCH")
print("=" * 80)

try:
    import torch

    print("Torch Version      :", torch.__version__)
    print("Torch CUDA         :", torch.version.cuda)
    print("CUDA Available     :", torch.cuda.is_available())
    print("GPU Count          :", torch.cuda.device_count())

    if torch.cuda.is_available():
        print("GPU Name           :", torch.cuda.get_device_name(0))
        print("cuDNN Version      :", torch.backends.cudnn.version())
        print("cuDNN Enabled      :", torch.backends.cudnn.enabled)

        x = torch.rand(1024, 1024).cuda()
        y = torch.rand(1024, 1024).cuda()
        z = x @ y

        print("Tensor Device      :", z.device)
        print("GPU Test           : PASS")
    else:
        print("GPU Test           : FAIL (CUDA unavailable)")

except Exception as e:
    print("PyTorch Error:", e)

print("\n" + "=" * 80)
print("OPENCV")
print("=" * 80)

try:
    import cv2
    print("OpenCV Version     :", cv2.__version__)
except Exception as e:
    print(e)

print("\n" + "=" * 80)
print("TENSORFLOW")
print("=" * 80)

try:
    import tensorflow as tf
    print("TensorFlow Version :", tf.__version__)
    print("GPUs               :", tf.config.list_physical_devices("GPU"))
except Exception as e:
    print(e)

print("\n" + "=" * 80)
print("GIT")
print("=" * 80)
os.system("git --version")

print("\n" + "=" * 80)
print("CUDA ENVIRONMENT")
print("=" * 80)
print("CUDA_VISIBLE_DEVICES =", os.environ.get("CUDA_VISIBLE_DEVICES"))
print("PATH =", os.environ.get("PATH"))

print("\n" + "=" * 80)
print("INSTALLED PACKAGES (Top 30)")
print("=" * 80)

try:
    import pkg_resources
    packages = sorted([(d.project_name, d.version) for d in pkg_resources.working_set])
    for name, version in packages[:30]:
        print(f"{name:<30} {version}")
    print(f"\nTotal Packages: {len(packages)}")
except Exception as e:
    print(e)

print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)

try:
    if torch.cuda.is_available():
        print("✅ GPU READY")
        print("GPU:", torch.cuda.get_device_name(0))
    else:
        print("❌ GPU NOT AVAILABLE")
except:
    print("❌ Unable to determine GPU status")
```
